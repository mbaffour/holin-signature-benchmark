Light notebook for poking at `holinbench` outputs after a `run-all`. Run the pipeline first:

```bash
holinbench run-all -c config/config.yaml
```

Then explore the result tables below. Remember: the bundled example data is **synthetic** and for pipeline validation only.

````python
import pandas as pd
from pathlib import Path

T = Path('..') / 'results' / 'tables'
def load(name):
    p = T / name
    return pd.read_csv(p, sep='\t') if p.exists() else pd.DataFrame()

sorted(p.name for p in T.glob('*.tsv'))
````
